```python
#latihan soal no 2

nama = input("nama barang :")
harga =int(input('harga satuan(Rp) :'))
jumlah =int(input("jumlah barang :"))
diskon =int(input("diskon (%) :"))
pajak =int(input("pajak(%) :"))
bayar =int(input("uang pembayaran(Rp) :"))

total= harga*jumlah
potongan = total * diskon//100
setelah_diskon = total - potongan
nilai_pajak = setelah_diskon * pajak //100
total_bayar = setelah_diskon + nilai_pajak
kembalian = bayar-total_bayar

print()
print("nama barang :",nama)
print("harga satuan : RP",harga)
print("jumlah :",jumlah)
print("total :Rp",total)
print("diskon", diskon, "% :Rp",potongan)
print("setelah diskon :Rp",setelah_diskon)
print("pajak",pajak,"% :Rp",nilai_pajak)
print("total bayar :Rp",total_bayar)
print("uang bayar :Rp",bayar)
print("kembalian :Rp",kembalian)

```

Budi membeli 7 buku dengan harga Rp 18.500 per buku. Toko memberi diskon 10% dari total belanja. Budi membayar dengan uang Rp 150.000.

Tentukan:

1. Total belanja, besar diskon, dan jumlah yang harus dibayar.
2. Uang kembalian.
3. Jika jumlah yang dibayar dipatungi 4 orang secara adil dalam bilangan bulat, berapa iuran per orang dan berapa rupiah sisanya?
4. Apakah uang Budi cukup? (hasil True/False)

```python
#soal 1
#4232501016_AHMAD FAUZAN

harga_buku= 18500
jumlah_buku= 7
diskon=10//100
uang_budi= 150000

#total belanja,besar diskon dan jumlah yang harus dibayar

total=harga_buku * jumlah_buku
diskon= total* 10//100
total_bayar= total-  diskon
print("total belanja:",total)
print("diskon 10 %:",diskon,"rupiah")
print("jumlah yang harus dibayar :",total_bayar,"rupiah")

#uang kembalian
kembalian= uang_budi-total_bayar
print("uang kembalian:",kembalian,"rupiah")

#jika jumlah yang dibayar dipatungi 4 orang secara adil dalam bilangan bulat,berapa iuran per orang dan berapa rupiah sisanya
ditraktir=int (total_bayar//4)
sisa_uang= int(total_bayar%4)
print("uang iuran per orang:",ditraktir,"rupiah")
print("uang sisanya:",sisa_uang,"rupiah")

#apakah uang budi cukup
cukup= uang_budi >= total
print("uang cukup:",cukup)
```

Sebuah rumah memasang 12 panel surya berdaya 450 Wp per panel. Rata-rata penyinaran efektif 5 jam/hari dengan efisiensi sistem 80% (0,8). Kebutuhan listrik rumah 25 kWh/hari dan tarif listrik Rp 1.444,70/kWh.

Tentukan:

1. Energi yang dihasilkan per hari (kWh).
2. Energi per bulan (30 hari) dan nilai rupiah listrik yang dihasilkan.
3. Kekurangan energi per hari terhadap kebutuhan.
4. Apakah PLTS sudah mencukupi kebutuhan rumah? (True/False)

```python
#4232501016_Ahmad Fauzan
#soal 2

daya_panel= 450
jumlah_panel= 12
penyinaran_efektif= 5
efisiensi= 0.8
kebutuhan_listrik= 25
tarif= 1444.70
 #energi yang dihasilkan per hari
total_daya= jumlah_panel*daya_panel/1000
energi_hari = total_daya * penyinaran_efektif * efisiensi
print("energi per hari :",energi_hari,"Kwh")

#energi perbulan dan nilai rupiah listrik yang dihasilkan

energi_bulan=energi_hari * 30
nilai_listrik= energi_bulan* tarif
print("energi per bulan :",energi_bulan,"Kwh")
print("nilai listrik :",nilai_listrik,"Rupiah")

#kekurangan energi per hari

kekurangan= kebutuhan_listrik - energi_hari
print("kekurangan energi:",round(kekurangan,2),"Kwh")

#apakah PLTS mencukupi kebutuhan

mencukupi= energi_hari>= kebutuhan_listrik
print("PLTS mencukupi:",mencukupi)
```

Sebuah generator diesel 200 kW beroperasi 8 jam/hari. Konsumsi bahan bakar 0,25 liter per kWh, harga solar Rp 13.500/liter, dan kapasitas tangki 1.000 liter. Kabel penghantar dialiri arus 20 A dengan hambatan 0,5 Ω.

Tentukan:

1. Energi yang dibangkitkan per hari (kWh).
2. Kebutuhan solar per hari dan biayanya.
3. Berapa hari penuh tangki cukup, dan berapa liter solar yang tersisa?
4. Rugi daya pada kabel (P = I² × R).
5. Apakah rugi kabel tergolong tinggi, yaitu lebih dari 150 W dan arus minimal 20 A? (True/False)

```python
#4232501016_Ahmad Fauzan
#soal 3

daya_generator= 200
jam_operasional= 8
konsumsi_solar=0.25
harga_solar= 13500
kapasitas_tangki=1000
arus= 20
hambatan= 0.5

#energi yang dibangkitkan per hari
energi_harian=daya_generator * jam_operasional
print("Energi (Kwh):",energi_harian)

#kebutuhan solar per hari dan biayanya

solar_hari= energi_harian*konsumsi_solar
biaya_hari= solar_hari * harga_solar
print("solar (liter):",solar_hari)
print("biaya (RP): ",biaya_hari)

#berapa hari penuh tangki cukup dan berapa liter sisanya

hari_penuh = kapasitas_tangki //solar_hari
hari_penuh = (hari_penuh)
solar_terpakai = hari_penuh * solar_hari
solar_sisa = kapasitas_tangki - solar_terpakai
print("Tangki cukup:", hari_penuh, "hari")
print("Sisa solar (L):", solar_sisa,)

#rugi daya pada kabel
rugi_daya= arus** 2 * hambatan
print("rugi kabel(W):",rugi_daya)

#apakah rugi kabel tergolong tinggi

tinggi = rugi_daya > 150 and arus >= 20
print("rugi tinggi? :",tinggi)
```

```python
total_belanja = 30000

if total_belanja > 20000:
  diskon = total_belanja * 0.10
  total_bayar= total_belanja - diskon
  print(f"anda mendapat diskon! total bayar : Rp {total_bayar}")
else:
  print(f"total bayar: Rp {total_belanja}")
```

If-elif- else
contoh penggunaan IF Majemuk

```python
nilai = 0
if nilai >=85:
  print("luar biasa! nilai A")
elif nilai >= 70:
  print("bagus! nilai B")
elif nilai >= 60:
  print("cukup. nilai C")
else:
  print("nilai kurang, perlu belajar lebih giat.")
```

```python
#menentukan jenis diskon berdasarkan hari

hari: str = "rabu"

if hari == "sabtu" or hari =="minggu":
  diskon: float = 0.20 #diskon 20%
  print(f"selamat! anda mendapat diskon akhir pekan sebesar {diskon* 100}%.")
elif hari == "jumat":
  diskon : float = 0.15 #diskon 15%
  print(f"diskon spesial jumat! anda mendapat potongan {diskon *100}%.")
else:
  diskon:float = 0.05 #diskon 5%
  print(f"anda mendapat diskon hari sebesar {diskon*100}%.")
```

```python
rupiah = int(input("memasukkan jumlah uang (Rp): "))
print("1. Dollar (USD)")
print("2. Euro (EUR)")
print("3. Yen (JPY)")
pilih = int(input("pilih konversi(1/2/3): "))

if pilih == 1:
  print("dalam Dollar:", rupiah/15000)
elif pilih == 2:
  print("dalam Euro:", rupiah/16000)
elif pilih == 3:
  print("dalam yen:", rupiah/120)
else:
  print("pilihan tidak valid")
```

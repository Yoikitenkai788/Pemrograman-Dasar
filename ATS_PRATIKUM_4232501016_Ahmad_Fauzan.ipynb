Sebuah PLTD beroperasi dengan daya_kw = 450 selama jam = 8 jam. Dalam sehari (24 jam) PLTD tersebut menghasilkan energi_aktual = 7200 kWh dari kapasitas_kw = 400.

a. Simpan nilai dalam variabel, hitung energi harian (energi = daya_kw × jam), lalu tampilkan dengan f-string. b. Hitung capacity factor: Capacity factor (%) = (energi_aktual / (kapasitas_kw × 24)) × 100. Tampilkan dengan 2 angka desimal.

Contoh output:

Energi harian = 3600 kWh
Capacity factor = 75.00%

```python
#nama = Ahmad Fauzan
#nim = 4232501016
#senin 9 November 2026
#pemrograman dasar
#soal no 2

daya_kw = 450
jam = 8
energi_aktual = 7200
kapasitas_kw = 400
# a
energi = daya_kw * jam
print(f"Energi harian = {energi} kWh")
#b
capacity_factor = (energi_aktual / (kapasitas_kw * 24)) * 100
print(f"Capacity factor = {capacity_factor:.2f}%")
```

cara kerja dari program ini pada bagian awal kita menyiapkan variabel yang dibutuhkan seperti daya kw, jam,energi aktual,kapasitas kw. pada soal a kita disuruh menghitung energi hariannya dengan cara daya kw dikali dengan jam  dan setelah itu tampilkan f stringnya. dari perhitungan tersebut didapat hasilnya 3.600 kwh
pada bagian b kita menghitung capacity factor dengan membagi energi aktual dengan kapasitas kw dikali 24 jam dan dikali 100 dengan catatan tampilan 2 angka desimal pada perhitungan tersebut didapatkan hasilnya 75.00 %

Status sebuah boiler ditentukan dari suhu uap (°C) dan tekanan (bar):

"STATUS: TRIP" jika suhu mencapai 550 atau lebih, atau tekanan mencapai 14 atau lebih.
"STATUS: WARNING" jika suhu lebih dari 520 dan tekanan lebih dari 12.
selain itu, "STATUS: NORMAL".
Baca suhu dan tekanan dari input (bertipe float). Gunakan if-elif-else dengan operator logika. Perhatikan urutan pemeriksaan kondisi.

```python
#nama = Ahmad Fauzan
#nim = 4232501016
#senin 9 November 2026
#pemrograman dasar

suhu = float(input("Masukkan suhu uap (°C): "))
tekanan = float(input("Masukkan tekanan (bar): "))
if suhu >= 550 or tekanan >= 14:
    print("STATUS: TRIP")
elif suhu > 520 and tekanan > 12:
    print("STATUS: WARNING")
else:
    print("STATUS: NORMAL")
```

Simulasikan pengosongan tangki dari 100 liter hingga kosong. Mulai dengan volume = 100. Gunakan perulangan while: selama volume > 0, kurangi 20 liter, lalu cetak volume saat ini (setelah dikurangi). Saat volume mencapai tepat 20 liter, cetak "WARNING: Hampir kosong!". Setelah loop selesai, cetak pesan tangki kosong.

```python
#nama = Ahmad Fauzan
#nim = 4232501016
#senin 9 November 2026
#pemrograman dasar

volume = 100
while volume > 0:
    volume -= 20
    print("Volume:", volume, "liter")
    if volume == 20:
        print("WARNING: Hampir kosong!")
print("TANGKI KOSONG - Pengosongan berhenti.")
```

cara kerja program ini dengan menyiapkan data volume yang diinginkan pada baris kedua kita menggunakan perintah while karena akan terus melakukan perulangan selama volume air masih lebih dari 0 didalam perulangan perintah if ditambahakan yaitu volume -20 digunakan untuk mengurangi volume air sebanyak 20 liter pada setiap tahap dan juga if volume == 20 berfungsi memeriksa apakah sisa air sudah mencapai 20 liter jika kondisi tersebut terpenuhi maka program akan menampilakn peringatan bahwa tangki hampir kosong Setelah volume air mencapai 0 liter, perulangan akan berhenti karena kondisi while volume > 0 sudah tidak terpenuhi. Terakhir, program menampilkan pesan "TANGKI KOSONG - Pengosongan berhenti." sebagai tanda bahwa proses pengosongan telah selesai.

soal 10

Data suhu bearing lima titik pengukuran dibaca sebagai teks, dan satu sensor rusak mengirim "N/A":

data = ["62.5", "58.0", "81.3", "N/A", "66.0"]

Periksa setiap data dengan for. Nomor titik dimulai dari 1. a. Ubah data menjadi float dengan try-except ValueError. Jika gagal, cetak Titik [n]: data tidak valid - pemeriksaan dihentikan, lalu hentikan perulangan dengan break. b. Jika nilai melebihi 70.0, cetak Warning: Titik [n] suhu bearing tinggi!. c. Hitung jumlah data valid yang berhasil diperiksa, dan tampilkan setelah perulangan selesai.

Output yang diharapkan untuk data di atas:

```python
#nama = Ahmad Fauzan
#nim = 4232501016
#senin 9 November 2026
#pemrograman dasar

data = ["62.5", "58.0", "81.3", "N/A", "66.0"]
jumlah_valid = 0
for i in range(len(data)):
    try:
        suhu = float(data[i])
        jumlah_valid += 1
        if suhu > 70.0:
            print(f"Warning: Titik {i+1} suhu bearing tinggi!")
    except ValueError:
        print(f"Titik {i+1}: data tidak valid - pemeriksaan dihentikan")
        break
print("Jumlah data valid:", jumlah_valid)
```

cara kerja Program dengan membuat daftar data suhu bearing dan mengatur jumlah data valid menjadi 0. Selanjutnya, program menggunakan perulangan for untuk memeriksa setiap data satu per satu. Data kemudian diubah menjadi angka menggunakan float dan dihitung sebagai data valid. Jika suhu melebihi 70°C, program akan menampilkan peringatan bahwa suhu bearing tinggi. Namun, jika ditemukan data yang tidak valid seperti N/A, program akan menampilkan pesan kesalahan dan menghentikan pemeriksaan menggunakan perintah break. Terakhir, program menampilkan jumlah data valid yang berhasil diperiksa, yaitu sebanyak 3 data.

Buatlah program yang menyelesaikan satu masalah di bidang energi (mis. pemantauan beban harian, deteksi anomali suhu, simulasi pengisian/pengosongan, atau penghitungan konsumsi listrik). Program minimal memuat:

percabangan if-elif-else,
perulangan for atau while,
sebuah list untuk menyimpan data.
atau dengan kombinasi lainya.
Tambahkan komentar yang menjelaskan masalah dan cara kerja program. Penggunaan try-except dan variabel/ekspresi yang tepat akan dinilai lebih baik.

```python
#nama = Ahmad Fauzan
#nim = 4232501016
#senin 9 November 2026
#pemrograman dasar

data_kwh = [12.5, 15.0, 22.5, 18.0, 28.0, 10.5, 25.0]
total = 0
for i in range(len(data_kwh)):
    try:
        energi = float(data_kwh[i])
        total += energi
        if energi < 15:
            status = "Rendah"
        elif energi <= 22:
            status = "Normal"
        else:
            status = "Tinggi"
        print(f"Hari ke-{i+1}: {energi} kWh - {status}")
    except ValueError:
        print(f"Hari ke-{i+1}: Data tidak valid")
rata_rata = total / len(data_kwh)
print("\nTotal konsumsi listrik:", total, "kWh")
print("Rata-rata konsumsi listrik:", round(rata_rata, 2), "kWh")
```

cara kerja program ini dengan membuat sebuah list yang berisi data konsumsi energi listrik selama 7 hari dan mengatur nilai total konsumsi menjadi 0. Selanjutnya, program menggunakan perulangan for untuk membaca dan memeriksa data konsumsi listrik setiap hari. Data tersebut kemudian diubah menjadi angka menggunakan float() agar dapat dihitung. Setelah itu, program menggunakan percabangan if-elif-else untuk mengelompokkan konsumsi listrik ke dalam kategori rendah, normal, atau tinggi berdasarkan jumlah energi yang digunakan. Jika terdapat data yang tidak valid, program akan menangani kesalahan menggunakan try-except. Selama proses pemeriksaan, setiap data yang valid akan dijumlahkan ke dalam variabel total. Terakhir, program menghitung dan menampilkan total serta rata-rata konsumsi energi listrik selama 7 hari.
